### Import necessary packages

```python
import pandas as pd
import sqlite3
```

### Loading cleaned dataset using Pandas 

```python
ola_data = pd.read_excel("ola_ride_cleaned_data.xlsx") 

print(ola_data.shape)
ola_data.head()
```

## Create a Database in SQLite

```python
conn_sqlite = sqlite3.connect("olaride_db.sqlite")  # Creates a database file
cursor = conn_sqlite.cursor()
print("SQLite connection established!")
```

## Create a Table in SQLite

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS ride_data (
        Date DATETIME,
    Time VARCHAR(20),
    Booking_ID VARCHAR(50) PRIMARY KEY,
    Booking_Status VARCHAR(50),
    Customer_ID VARCHAR(50),
    Vehicle_Type VARCHAR(50),
    Pickup_Location VARCHAR(100),
    Drop_Location VARCHAR(100),
    Canceled_Rides_by_Customer VARCHAR(100),
    Canceled_Rides_by_Driver VARCHAR(100),
    Incomplete_Rides VARCHAR(100),
    Incomplete_Rides_Reason VARCHAR(255),
    Booking_Value INT,
    Payment_Method VARCHAR(50),
    Ride_Distance INT,
    Driver_Ratings FLOAT,
    Customer_Rating FLOAT,
    Vehicle_Images VARCHAR(255),
    year INT,
    month INT,
    day INT,
    weekday VARCHAR(20)
)
""")
conn_sqlite.commit()

print("Table 'ride_data' created successfully in SQLite!")
```

## Data Insertion

```python
ola_data.to_sql(
    "ride_data",
    conn_sqlite,
    if_exists="replace",
    index=False
)

print("Data inserted successfully!")
```
